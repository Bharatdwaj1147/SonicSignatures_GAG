```python
import librosa
import librosa.display
import matplotlib.pyplot as plt
import numpy as np

songs = ["song1.mp3", "song2.mp3", "song3.mp3"]

for song in songs:
    audio, sr = librosa.load(song)

    # Plot Waveform
    plt.figure(figsize=(10, 4))
    librosa.display.waveshow(audio, sr=sr, color="blue")
    plt.title(f"Waveform: {song}")
    plt.xlabel("Time (s)")
    plt.ylabel("Amplitude")
    plt.tight_layout()
    plt.show()

    # Compute STFT and Plot Spectrogram (in dB)
    spectrogram = librosa.amplitude_to_db(np.abs(librosa.stft(audio)))
    plt.figure(figsize=(10, 4))
    librosa.display.specshow(spectrogram, sr=sr, x_axis="time", y_axis="log")
    plt.colorbar(format="%+2.0f dB")
    plt.title(f"Spectrogram: {song}")
    plt.tight_layout()
    plt.show()
```
